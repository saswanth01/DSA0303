```python
import nltk
from nltk import word_tokenize, pos_tag

# Download necessary NLTK data resources
nltk.download('punkt_tab') # As suggested by the error message
nltk.download('averaged_perceptron_tagger_eng') # Corrected: Required for pos_tag for English

sentence = "The experienced engineer carefully designed the bridge."

tokens = word_tokenize(sentence)
tags = pos_tag(tokens)

print(tags)
```
