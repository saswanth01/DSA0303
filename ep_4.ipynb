```python
import nltk
from nltk import word_tokenize
from nltk.util import ngrams
from nltk import CFG
from nltk.parse import ChartParser
from collections import Counter

# Download tokenizer
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve the LookupError

# -----------------------------
# PART A : UNSMOOTHED N-GRAM MODEL
# -----------------------------

corpus = """
the cat sat on the mat
the cat ate fish
the dog sat on the mat
the dog ate bone
"""

tokens = word_tokenize(corpus.lower())

print("\nTokens:\n")
print(tokens)

# Unigrams
unigrams = Counter(tokens)

# Bigrams
bigrams = Counter(ngrams(tokens,2))

# Trigrams
trigrams = Counter(ngrams(tokens,3))

print("\nUnigram Counts\n")
print(unigrams)

print("\nBigram Counts\n")
print(bigrams)

print("\nTrigram Counts\n")
print(trigrams)

# -----------------------------
# Probabilities
# -----------------------------

total_words = len(tokens)

print("\nUnigram Probabilities\n")

for word,count in unigrams.items():
    print(word,"=",round(count/total_words,4))

print("\nBigram Probabilities\n")

for bg,count in bigrams.items():
    probability = count/unigrams[bg[0]]
    print(bg,"=",round(probability,4))

print("\nTrigram Probabilities\n")

for tg,count in trigrams.items():

    previous_bigram=(tg[0],tg[1])

    probability=count/bigrams[previous_bigram]

    print(tg,"=",round(probability,4))

# -----------------------------
# Check unseen bigram
# -----------------------------

query=("cat","runs")

if query in bigrams:

    probability=bigrams[query]/unigrams[query[0]]

else:

    probability=0

print("\nProbability of",query,"=",probability)

# -----------------------------
# PART B : CFG PARSE TREE
# -----------------------------

grammar = CFG.fromstring("""

S -> NP VP

NP -> Det N

VP -> V NP

Det -> 'the'

N -> 'cat' | 'dog'

V -> 'chased'

""")

parser = ChartParser(grammar)

sentence = "the cat chased the dog".split()

print("\nParse Tree\n")

for tree in parser.parse(sentence):

    print(tree)

    tree.pretty_print()
```
